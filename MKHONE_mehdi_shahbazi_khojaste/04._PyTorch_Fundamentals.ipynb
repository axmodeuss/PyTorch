##### **Alternative ways to convert data to tensor**

```python
import numpy as np
np_array = np.array([1, 2, 3, 4, 5], dtype=np.float64)
```

###### **1. torch.from_numpy**

```python
# 1. torch.from_numpy() |--> shares memory with np_array, so changes in np_array reflect in tensor_from_numpy
tensor_from_numpy = torch.from_numpy(np_array)

print("Original NumPy array:", np_array)
print("torch.from_numpy:", tensor_from_numpy)
```

```python
np_array[0] = 10 # Modify the NumPy array
print("Modified NumPy array:", np_array)
print("torch.from_numpy:", tensor_from_numpy)
```

###### **2. torch.as_tensor**

```python
# 2. torch.as_tensor() |--> Copies if NumPy but If already a tensor, no copy is made.
# 2.1 --> input is NumPy array
np_array = np.array([1, 2, 3, 4, 5], dtype=np.float64)

tensor_as_tensor = torch.as_tensor(np_array)
print("Original NumPy array:", np_array)
print("torch.as_tensor:", tensor_as_tensor)
```

```python
np_array[1] = 20
print("Modified NumPy array:", np_array)
print("Tensor after modifying NumPy array:", tensor_as_tensor)
```

```python
# 2.2 --> input is Tensor with same dtype
input_tensor = torch.arange(0, 5, dtype=torch.int16)
tensor_as_tensor = torch.as_tensor(input_tensor)

tensor_as_tensor[-2] = 50
print("Input Tensor:", input_tensor)
print("Tensor after modifying:", tensor_as_tensor)
```

```python
# 2.3 --> input is Tensor with different dtype
input_tensor = torch.arange(0, 5, dtype=torch.int16)
tensor_as_tensor = torch.as_tensor(input_tensor, dtype=torch.float64)

tensor_as_tensor[-2] = 50
print("Input Tensor:", input_tensor)
print("Tensor after modifying:", tensor_as_tensor)
```

```python
# 2.4 --> input is List
list_data = [1, 2, 3, 4, 5]
tensor = torch.as_tensor(list_data)

# Modify the list
list_data[0] = 10
print("Modified list:", list_data)
print("Tensor after modifying list:", tensor)
```

###### **3. torch.tensor**

```python
# 3. torch.tensor() |--> Always creates a new tensor and does not share memory with the source data; a new copy is made.
np_array = np.array([1, 2, 3, 4, 5])

tensor_np_array = torch.tensor(np_array)
list_data = [1, 2, 3, 4, 5]
tensor_from_list = torch.tensor(list_data)
```

```python
tensor_np_array[-2] = 100
print("Original NumPy array:", np_array)
print("Tensor after modifying:", tensor_np_array)
```

```python
list_data[-2] = 100

print("Original list:", list_data)
print("Tensor after modifying:", tensor_from_list)
```

##### **Automatic differentiation**

The most important advantage of PyTorch over NumPy is its automatic differentiation functionality which is very useful in optimization applications such as optimizing parameters of a neural network. Let's try to understand it with an example.

Say you have a composite function which is a chain of two functions: `g(u(x))`.
To compute the derivative of `g` with respect to `x` we can use the chain rule which states that: `dg/dx = dg/du * du/dx`. PyTorch can analytically compute the derivatives for us.

To compute the derivatives in PyTorch first we create a tensor and set its `requires_grad` to true. We can use tensor operations to define our functions. We assume `u` is a quadratic function and `g` is a simple linear function:

```python
x = torch.tensor(1.0, requires_grad=True)

def u(x):
  return x * x

def g(u):
  return -u
```

In this case our composite function is `g(u(x)) = -x*x`. So its derivative with respect to `x` is `-2x`. At point `x=1`, this is equal to `-2`.

Let's verify this. This can be done using grad function in PyTorch:

```python
dgdx = torch.autograd.grad(g(u(x)), x)[0]
print(dgdx)  # tensor(-2.)
```

##### **Matrix Multiplication**

```python
# Matrix multiplication functionality is in the torch.matmul() method.

my_tensor = torch.tensor([1, 2, 3])

print("Element-wise multiplication:", my_tensor * my_tensor, "| Math: [1*1, 2*2, 3*3] = [1, 4, 9]")
print()
print("Matrix multiplication:", torch.matmul(my_tensor, my_tensor), "| Math: [1*1 + 2*2 + 3*3] = [14]")
```

```python
tensor_a = torch.rand(4, 8)
tensor_b = torch.rand(8, 4)

print("Shapes:")
print("tensor_a:", tensor_a.shape, "|", "tensor_b:", tensor_b.shape)
```

```python
matrix_multiplication = torch.matmul(tensor_a, tensor_b)
print('Input Shape: ', tensor_a.shape, '|', tensor_b.shape)
print('Output Shape: ', matrix_multiplication.shape)
```

```python
matrix_multiplication = torch.matmul(tensor_b, tensor_a)
print('Input Shape: ', tensor_b.shape, '|', tensor_a.shape)
print('Output Shape: ', matrix_multiplication.shape)
```

```python
try:
  torch.matmul(tensor_a, tensor_a)
except Exception as err:
  print("Error: ", err)
  print()
  print('Input Shape: ', tensor_a.shape, '|', tensor_a.shape)
```

```python
# One way to solve a shape mismatch is by using a transpose operation (switching the dimensions of a given tensor).

print('Shape of transpose of tensor_a: ', tensor_a.T.shape)
print()

transpose = tensor_a.T
torch.matmul(tensor_a, transpose)
```

```python
# Other shortcuts for torch.matmul

a = tensor_a.matmul(transpose)
b = torch.mm(tensor_a, transpose)

a == b
```

###### **Stacking**

```python
x = torch.arange(0, 6, 1)
print(f'x value: {x}, | x shape: {x.shape}')
```

```python
# Stack tensors on top of each other
stacked = torch.stack([x, x, x], dim=0) # dim = 0
print(f'stacked value:\n {stacked}')
print()
print(f'stacked shape: {stacked.shape}')
```

```python
# Stack tensors on top of each other
stacked = torch.stack([x, x, x], dim=1) # dim = 1
print(f'stacked value:\n {stacked}')
print()
print(f'stacked shape: {stacked.shape}')
```

# Seeding (Reproducibility)

###### **With Numpy**

```python
def random_number_generator(seed=None):
    if seed:
        np.random.seed(seed=123)

    random_list = []
    for i in range(6):
        random_number = np.float16(np.random.rand())
        random_list.append(random_number)

    return random_list

print('List A: ', random_number_generator())
print('List B: ', random_number_generator())
print('List C: ', random_number_generator())
```

```python
# Repeat the same process with seeding
my_seed = 123

print('List A: ', random_number_generator(seed=my_seed))
print('List B: ', random_number_generator(seed=my_seed))
print('List C: ', random_number_generator(seed=my_seed))
```

###### **With Torch**

```python
def random_torch(seed):
    torch.random.manual_seed(seed)
    random_tensor = torch.rand(3, 4)

    return random_tensor

Tensor_A = random_torch(seed=my_seed)
Tensor_B = random_torch(seed=my_seed)

print('Tensor A:\n', Tensor_A)
print()
print('Tensor B:\n', Tensor_B)
print()
print(f"Does Tensor A equal Tensor B?")
print(Tensor_A == Tensor_B)
```
