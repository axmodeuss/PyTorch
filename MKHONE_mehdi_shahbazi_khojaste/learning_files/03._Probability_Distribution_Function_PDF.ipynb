```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

# Probability Distribution Function (PDF)

Normal distribution (Gaussian distribution)

$\Large f(x | \mu, \sigma) = \frac{1}{\sqrt{2\pi} \cdot \sigma} e^{-\frac{1}{2}(\frac{x - \mu}{\sigma})^2} $

```python
x = np.linspace(-4, 4, 1000)

mu = 0
sigma = 1
pdf_values = 1 / (np.sqrt(2 * np.pi) * sigma) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)


plt.figure(figsize=(8, 6))
plt.plot(x, pdf_values, label='Normal Distribution')


plt.xlabel('x')
plt.ylabel('Probability Density')
plt.title('Normal (Gaussian) Distribution')

plt.grid(True)
plt.legend()

plt.show()
```

```python
data = np.random.normal(loc=5, scale=1, size=10_000_000) # Sampling from distribution
sns.displot(data, kind='kde') # kernel density estimation

plt.show()
```

Uniform distribution

```python
data = np.random.uniform(size=10_000_000)
sns.displot(data, kind='kde')

plt.show()
```
