```python
tesns1 = torch.rand(size=(2,3),device="cuda")
tesns2 = torch.rand(size=(2,3),device="cuda")
tesns2_prime = torch.transpose(tesns2,0,1)
print(tesns1)
print(tesns2)
print(tesns2_prime)
```

```python
mat_tesns = tesns1@tesns2_prime
print(mat_tesns)
```

```python
#7 MinMax
print("Max : ",torch.max(mat_tesns))
print("Min : ",torch.min(mat_tesns))
```

```python
#8 argmax and argmin
arg_max = torch.argmax(mat_tesns)
arg_min = torch.argmin(mat_tesns)
print(arg_max,arg_min)
```
