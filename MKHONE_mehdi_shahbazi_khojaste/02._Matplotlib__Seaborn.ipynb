```python
import numpy as np
import matplotlib.pyplot as plt
```

# Line Plot

Single line plot

```python
x = np.array([0, 3, 6])
y = np.array([0, 50, 250])

plt.plot(x, y)
plt.show()
```

```python
value = np.array([3, 8, 1, 10, 5, 7])

plt.plot(value, marker = '*', linestyle='-.', color='black')
plt.show()
```

```python
x = np.arange(0, 5 * np.pi, 0.1)
y = np.sin(x)

plt.plot(x, y)
plt.grid()
plt.show()
```

Double line plot

```python
y_sin = np.sin(x)
y_cos = np.cos(x)


plt.plot(x, y_sin)
plt.plot(x, y_cos)

plt.xlabel('x')
plt.ylabel('y')

plt.title('Sine and Cosine')

plt.legend(['Sine', 'Cosine'])

plt.show()
```

Sub plot

```python
plt.subplot(2, 1, 1)
plt.plot(x, y_sin)
plt.title('Sine')

plt.subplot(2, 1, 2)
plt.plot(x, y_cos)
plt.title('Cosine')

plt.show()
```

# Scatter Plot

```python
x = np.random.randint(15, size=15)
y_1 = np.random.randint(low=85, high=100, size=15)

plt.scatter(x, y_1, s=150, color='hotpink')


y_2 = np.random.randint(low=85, high=100, size=15)
plt.scatter(x, y_2, s=350, color='#70c999')
plt.show()
```

Rainbow

```python
x = np.random.randint(100, size=100)
y = np.random.randint(100, size=100)
colors = np.random.randint(100, size=(100))
sizes = 10 * np.random.randint(100, size=(100))

plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='nipy_spectral')

plt.colorbar()

plt.show()
```

# Bar Chart

Vertical

```python
x = np.array(["A", "B", "C", "D"])
y = np.array([5, 14, 3, 18])

plt.bar(x, y, color = "hotpink")
plt.show()
```

Horizontal

```python
plt.barh(x, y, height = 0.3, color = "#4CAF50")
plt.show()
```

# Image Show

Image

```python
from PIL import Image
img = Image.open('unnamed.jpg')

plt.imshow(img)
plt.show()
```

Colormap - without aspect

```python
gradient = np.linspace(0, 1, 256)

gradient = np.vstack((gradient, gradient))

plt.imshow(gradient, cmap='nipy_spectral')
plt.show()
```

Colormap - with aspect

```python
plt.imshow(gradient, aspect='auto', cmap='nipy_spectral')
plt.axis('off')
plt.title('Colormap')
plt.show()
```

# Histogram

```python
x = np.random.normal(loc=100, scale=2, size=300) # loc = mean | scale = standard deviation

plt.hist(x, color='teal', edgecolor='black')
plt.show()
```

# Pie Chart

```python
y = np.array([5, 10, 25, 45])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]

plt.pie(y, labels=None, colors=plt.cm.plasma_r(np.linspace(0, 1, len(y))))
plt.legend(mylabels, loc=[1, 0.5])
plt.show()
```

# Seaborn

```python
import seaborn as sns
```

Plot pairwise relationships

```python
data = sns.load_dataset("iris")
sns.pairplot(data=data, hue='species')

plt.show()
```

Regression plot

```python
sns.lmplot(x='sepal_length', y='petal_width', data=data)
plt.show()
```
